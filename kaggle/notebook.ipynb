```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../dataset/press_release_benchmarks.csv')
print('Dataset shape:', df.shape)
print(df.head(10))
```

```python
signal_cols = ['pr_quality_score','publication_readiness_score','seo_optimization_score','tech_keywords_score','media_distribution_score','newsworthiness_score']
signal_avgs = df[signal_cols].mean()
colors = ['#3b82f6','#10b981','#f59e0b','#8b5cf6','#ef4444','#06b6d4']
plt.figure(figsize=(12,5))
plt.bar(signal_avgs.index, signal_avgs.values, color=colors)
plt.title('TechBullion - Average PR Signal Scores', fontsize=14, fontweight='bold')
plt.xlabel('PR Signal')
plt.ylabel('Average Score (0-100)')
plt.xticks(rotation=20)
plt.axhline(y=75, color='green', linestyle='--', alpha=0.5, label='Healthy threshold')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
type_counts = df['content_type'].value_counts()
plt.figure(figsize=(10,5))
plt.bar(type_counts.index, type_counts.values, color=colors[:len(type_counts)])
plt.title('TechBullion - Content Type Distribution', fontsize=14, fontweight='bold')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```
